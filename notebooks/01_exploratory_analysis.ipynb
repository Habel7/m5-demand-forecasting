## 1. Intermittency — the defining feature of retail demand

TODO: compute the share of zero-sales days per item and plot the distribution.

```python
# TODO: pick the d_ columns, compute (sales[d_cols] == 0).mean(axis=1),
# The d_ columns hold daily unit sales; grab them
d_cols = [c for c in sales.columns if c.startswith("d_")]

# For each item, what fraction of days had zero sales?
zero_rate = (sales[d_cols] == 0).mean(axis=1)

print(f"Median item sells 0 units on {zero_rate.median():.0%} of days")

zero_rate.hist(bins=50)
plt.xlabel("Fraction of days with zero sales")
plt.ylabel("Number of items")
plt.title("Most SKUs are intermittent sellers")
plt.savefig("../results/figures/intermittency.png", dpi=120, bbox_inches="tight")
plt.show()
```

This means that the median SKU sells zero units on 74% of days, and the distribution clusters near 1.0, meaning a large share of items sell nothing on more than 90% of days. This matters because a model that predicts each item's average would be wrong almost every day, since it can never output the zeros that dominate, nor the occasional spike when a unit actually sells. It means the project needs a proper error metric like WRMSSE, which weights items by how much they actually sell rather than plain accuracy, and an event-aware model that can capture demand spikes around promotions and holidays instead of predicting a flat average. From my own experience on a supermarket floor, these slow-moving items are where a company generates the most reduced to clear waste and thereby a significant loss.

## 2. Seasonality at category level

TODO: melt to long format (see `melt_sales`), aggregate daily units by `cat_id`, plot FOODS vs HOUSEHOLD vs HOBBIES over time. Then a day-of-week profile.

```python
# long = melt_sales(sales, calendar)  # ~60M rows — filter to one store first if RAM is tight
# Sum daily units across all items, grouped by category
d_cols = [c for c in sales.columns if c.startswith("d_")]

cat_daily = sales.groupby("cat_id")[d_cols].sum()   # 3 rows (categories) x 1913 days

# Transpose so days are rows, then attach real dates from the calendar
cat_daily = cat_daily.T                               # days become rows
cat_daily.index = calendar.set_index("d").loc[cat_daily.index, "date"].values

# Plot the three categories over time
cat_daily.plot(figsize=(14, 5))
plt.title("Daily units sold by category over time")
plt.ylabel("Total units sold")
plt.xlabel("Date")
plt.savefig("../results/figures/category_trend.png", dpi=120, bbox_inches="tight")
plt.show()
```

When aggregated to category level, FOODS dominates the volume at roughly 25,000 units/day on average, peaking near 35,000, which is about 5× higher than HOUSEHOLD and HOBBIES. The sharp annual dips to zero are the Christmas Day closures, when the stores shut. There is also a mild increasing trend across the five years, along with dense weekly fuzz on each line. This tells me the model needs day of week and holiday features, because the demand is clearly structured rather than random. On the floor, FOODS is daily replenishment stock, which is why it carries the volume here.

```python
# Add weekday to the daily category table
cat_daily["weekday"] = pd.to_datetime(cat_daily.index).day_name()

order = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
dow = cat_daily.groupby("weekday")[["FOODS","HOUSEHOLD","HOBBIES"]].mean().reindex(order)

dow.plot(kind="bar", figsize=(12, 5))
plt.title("Average units sold by day of week")
plt.ylabel("Average units")
plt.xticks(rotation=45)
plt.savefig("../results/figures/category_dow.png", dpi=120, bbox_inches="tight")
plt.show()
```

The average daily sales climb into the weekend. Saturday and Sunday reach about 30,000 units for FOODS against ~22,000 midweek, which is a lift of roughly 25–30%. HOUSEHOLD follows the same shape, while HOBBIES stays flatter because hobby purchases are discretionary and one off rather than part of the routine weekly shop. There is a small Monday bump and a midweek trough on Wednesday and Thursday. This is direct evidence that day of week must be a model feature. From working supermarket shifts, this is the same pattern behind more staff being rostered and larger orders going in on weekends.
